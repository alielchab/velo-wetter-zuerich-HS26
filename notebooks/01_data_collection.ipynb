# Project: Impact of Weather on Bicycle Traffic in Zurich

### Goal
In this notebook, bicycle traffic count data from the City of Zurich for the years 2023–2025 are collected and analyzed for the first time.

### Data Collection
- **Provider**: Open Data Zurich, Public Works Department of the City of Zurich
- **Dataset**: Data from automatic pedestrian and bicycle counts - quarter-hourly values
- **Time period**: 2023–2025
- **Format**: CSV
- **Source**: https://data.stadt-zuerich.ch/dataset/ted_taz_verkehrszaehlungen_werte_fussgaenger_velo

### Procedure
1. Import Python libraries
2. Download raw data and save it locally
3. Check the data structure
4. Perform an initial data quality review



```python
# import all necessary libraries

import pandas as pd
import requests
from pathlib import Path

print("Alle Bibliotheken erfolgreich importiert!")
```

```python

# determine projectfolder 
project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

# Ordner für Originaldaten
raw_dir = project_root / "data" / "raw"

# Sicherstellen, dass der Ordner existiert
raw_dir.mkdir(parents=True, exist_ok=True)

print("Projektordner:", project_root)
print("Speicherort:", raw_dir)
```

```python
import requests
from pathlib import Path

year = 2025

# Download URL
url = (
    "https://data.stadt-zuerich.ch/dataset/ted_taz_verkehrszaehlungen_werte_fussgaenger_velo/"
    f"download/{year}_verkehrszaehlungen_werte_fussgaenger_velo.csv"
)

# Save path
raw_dir.mkdir(parents=True, exist_ok=True)
file_path = raw_dir / f"velo_{year}.csv"

# Download if file is missing or empty
if not file_path.exists() or file_path.stat().st_size == 0:

    print("Downloading started...")

    with requests.get(url, stream=True, timeout=60) as response:
        response.raise_for_status()

        # IMPORTANT: Keep the file open inside the request block
        with open(file_path, "wb") as file:
            for chunk in response.iter_content(chunk_size=1024*1024):
                if chunk:
                    file.write(chunk)

    # Check downloaded file size
    file_size = file_path.stat().st_size

    if file_size == 0:
        file_path.unlink()
        raise ValueError("Download failed: CSV file is empty!")

    print("Download successful!")
    print(f"File size: {file_size / (1024**2):.2f} MB")

else:
    print("File already exists:", file_path)
```
