```python

import pandas as pd

url = (
    "https://data.stadt-zuerich.ch/dataset/"
    "ted_taz_verkehrszaehlungen_werte_fussgaenger_velo/"
    "download/2025_verkehrszaehlungen_werte_fussgaenger_velo.csv"
)

# Zuerst eine Stichprobe laden
df = pd.read_csv(url, nrows=10000)

print("Anzahl Zeilen:", len(df))
print("Spalten:", df.columns.tolist())
print(df.head())

# Datentypen vorbereiten
df["DATUM"] = pd.to_datetime(
    df["DATUM"], errors="coerce"
)

for col in ["VELO_IN", "VELO_OUT"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Gesamtzählungen pro Messintervall
df["velo_total"] = df[
    ["VELO_IN", "VELO_OUT"]
].sum(axis=1, min_count=1)

print(df.isna().sum())
```
